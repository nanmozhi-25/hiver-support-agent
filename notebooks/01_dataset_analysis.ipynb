# 01. Dataset Inspection and Brand Analysis
This notebook inspects the Twitter Customer Support dataset (`twcs.csv`), computes per-brand conversation volume, evaluates message quality/noise, ranks candidate brands, and justifies selecting **AmazonHelp** as the primary target brand.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot styles
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Load Brand Analysis Metrics
Load the precomputed brand analysis produced by `src/data_loader.py`.

```python
df_analysis = pd.read_csv('../data/brand_analysis.csv')
df_analysis.head(10)
```

## 2. Visualize Top Candidate Brands by Conversation Volume

```python
plt.figure(figsize=(12, 6))
top_10 = df_analysis.head(10)
sns.barplot(data=top_10, x='est_conversations', y='brand', palette='viridis')
plt.title('Top 10 Support Brands by Multi-Turn Conversation Count')
plt.xlabel('Estimated Conversations')
plt.ylabel('Brand Handle')
plt.tight_layout()
plt.savefig('../results/figures/brand_conversations.png')
plt.show()
```

## 3. Brand Selection Justification
**Selected Brand: AmazonHelp**

**Justification Factors:**
1. **Volume & Depth**: 100,503 customer conversations (over 2.8x more than the #2 candidate, AppleSupport).
2. **Text Richness**: Average message length of 19.26 words, providing sufficient context for intent classification and retrieval.
3. **Data Quality**: Low noise profile with < 1.3% duplicate text and 0 missing values.
4. **Problem Diversity**: Encompasses wide customer support domain: shipping/delivery tracking, refunds, returns, order cancellation, Prime subscriptions, account access, defective items, and billing.
